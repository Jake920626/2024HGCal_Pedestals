```python
import ROOT
import numpy as np

# 開啟 root 檔案
file = ROOT.TFile.Open("./Relay1727132796.root","read")

# 列出所有 TGraphErrors 的名字
keys = file.GetListOfKeys()

arr = np.array([])
all_data = []

# 遍歷所有 TGraphErrors
for key in keys:
    obj = key.ReadObj()
    if isinstance(obj, ROOT.TGraphErrors):
        graph = obj
        name = graph.GetName()

        n = graph.GetN()
        for i in range(n):
            ey = graph.GetErrorY(i)
            arr = np.append(arr, ey)

        all_data.append(arr)
        arr = np.array([])
        
print(type(ey))


      
```

```python
all_data = np.array(all_data)
```
